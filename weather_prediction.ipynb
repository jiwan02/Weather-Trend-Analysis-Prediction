```python
cols = list(df.columns)
cols
```

```python
import re
split_cols = []
for col in cols:
  match = re.match(r"([A-Z_]+)_(.*)", col)
  if match:
    city, variable = match.groups()
    split_cols.append([city, variable])

mapping_df = pd.DataFrame(split_cols, columns=["City", "WeatherVariable"])
print(mapping_df)
```

```python
# put city name on column1 city then weather variable in after column1
# helper column that just repeats the variable name
mapping_df["Value"] = mapping_df["WeatherVariable"]
summary = mapping_df.pivot_table(
    index="City",
    columns="WeatherVariable",
    values="Value",
    aggfunc=lambda x: list(x)[0],
    fill_value=""
)

summary = summary.fillna("")

summary["Present_Count"] = (summary != "").sum(axis=1)
summary["Missing_Count"] = (summary == "").sum(axis=1)

pd.set_option('display.max_columns', None)
print(summary)
```

```python
incomplete_cities = summary[summary["Missing_Count"]>=1]
print(incomplete_cities)
```

```python
incomplete_city_list = incomplete_cities.index.tolist()
print(incomplete_city_list)
```

```python
#Drop all columns that start with those city names
import pandas as pd
df_copy = df.copy()
df_copy = df_copy.drop(
    columns=[col for col in df_copy.columns
             if any(col.startswith(city + "_") for city in incomplete_city_list)]
)

pd.set_option('display.max_columns', None)
print(df_copy.head().to_string())
```

```python
df_copy.columns
```

```python
# Keep only columns that end with "temp_mean" or "humidity"
keep_cols = [col for col in df_copy.columns
             if col.endswith("temp_mean") or col.endswith("humidity")]
keep_cols = ["DATE", "MONTH"] + keep_cols
df_copy = df_copy[keep_cols]

pd.set_option('display.max_columns', None)
print(df_copy.head().to_string())
```

```python
plt.figure(figsize=(12,6))   # wide plot
sns.heatmap(df_copy.corr(), annot=True, cmap="coolwarm")
plt.show()


#df_copy['DATE'] = pd.to_datetime(df_copy['DATE'].astype(str).str.extract(r'(\d{8})')[0])
df_copy['DATE'] = pd.to_datetime(df_copy['DATE'], format='%Y%m%d')

pd.set_option('display.max_columns', None)
print(df_copy.head().to_string())
print(df_copy['DATE'].dtype)
df_copy.info()
print(df_copy.dtypes)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Select first 5 rows for demo
df = df_copy.head(5)

cities = ['DE_BILT', 'DUSSELDORF', 'MAASTRICHT', 'MUENCHEN', 'OSLO']
temp_cols = [c + '_temp_mean' for c in cities]
hum_cols  = [c + '_humidity' for c in cities]

# Average values across the sample
temp_means = [df[col].mean() for col in temp_cols]
hum_means  = [df[col].mean() for col in hum_cols]

x = np.arange(len(cities))  # city positions
width = 0.35

plt.figure(figsize=(10,6))
plt.bar(x - width/2, temp_means, width, label='Temp Mean')
plt.bar(x + width/2, hum_means, width, label='Humidity')

plt.xticks(x, cities, rotation=45)
plt.ylabel('Values')
plt.title('Comparative Weather Metrics (Temp vs Humidity)')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,6))
for col in temp_cols:
    plt.hist(df_copy[col], bins=20, alpha=0.5, label=col)

plt.title('Temperature Distribution Across Cities')
plt.legend()
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10,8))
sns.heatmap(df_copy[temp_cols + hum_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Between Cities (Temp & Humidity)')
plt.show()
```

```python
import pandas as pd

def classify_weather(temp, humidity):
    if humidity > 0.85 and temp < 20:
        return "Rainy"
    elif 0.6 <= humidity <= 0.85:
        return "Cloudy"
    else:
        return "Sunny"

df_copy['Weather'] = df_copy.apply(
    lambda row: classify_weather(row['DE_BILT_temp_mean'], row['DE_BILT_humidity']),
    axis=1
)

print(df_copy[['DATE','DE_BILT_temp_mean','DE_BILT_humidity','Weather']].head(50))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df_copy[['DE_BILT_temp_mean','DE_BILT_humidity']]
y = df_copy['Weather']   # labels created by rule

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)

print("Accuracy:", model.score(X_test, y_test))
```
