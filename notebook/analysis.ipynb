<h1>Bar Graph visualization</h1>

1. City vs Average Temperature

```python
avg_temp = [
df['BASEL_temp_mean'].mean(),
df['STOCKHOLM_temp_mean'].mean(),
df['TOURS_temp_mean'].mean()
]

cities = ['Basel','Stockholm','Tours']

plt.figure(figsize=(8,5))
plt.bar(cities, avg_temp)

plt.title("Average Temperature Comparison Across Cities")
plt.xlabel("City")
plt.ylabel("Average Temperature")

plt.show()
```

Outcome

The bar graph compares average temperature across cities.

The city with the highest bar has the highest average temperature.

This helps identify which city is warmer or cooler overall.

2.City vs Average Precipitation

```python
avg_rain = [
df['BASEL_precipitation'].mean(),
df['STOCKHOLM_precipitation'].mean(),
df['TOURS_precipitation'].mean()
]

cities = ['Basel','Stockholm','Tours']

plt.figure(figsize=(8,5))
plt.bar(cities, avg_rain)

plt.title("Average Precipitation Comparison Across Cities")
plt.xlabel("City")
plt.ylabel("Average Precipitation")

plt.show()
```

Outcome

The bar graph shows which city receives more rainfall on average.

A taller bar indicates higher rainfall frequency or intensity.

3.City vs Average Sunshine

```python
avg_sun = df['BASEL_sunshine'].mean()

plt.figure(figsize=(6,4))
plt.bar(['Basel'], [avg_sun])

plt.title("Average Sunshine Hours in Basel")
plt.xlabel("City")
plt.ylabel("Average Sunshine")

plt.show()
```

Outcome

Shows average sunshine hours.

4.Monthly Average Temperature per City

```python
df['DATE'] = pd.to_datetime(df['DATE'])

df['month'] = df['DATE'].dt.month

monthly_temp = df.groupby('month')[[
'BASEL_temp_mean',
'STOCKHOLM_temp_mean',
'TOURS_temp_mean'
]].mean()

monthly_temp.plot(kind='bar', figsize=(10,6))

plt.title("Monthly Average Temperature per City")
plt.xlabel("Month")
plt.ylabel("Average Temperature")

plt.show()
```

Outcome

Shows temperature variation across months.

Helps identify seasonal patterns.

<h1>Pie Chart

Pie Chart — Rainy vs Non-Rainy Days

```python
rainy_days = (df['BASEL_precipitation'] > 0).sum()
non_rainy_days = (df['BASEL_precipitation'] == 0).sum()

labels = ['Rainy Days', 'Non-Rainy Days']
values = [rainy_days, non_rainy_days]

plt.figure(figsize=(6,6))

plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=90)

plt.title("Rainy vs Non-Rainy Days (Basel)")

plt.show()
```

Outcome

The pie chart shows the proportion of rainy and non-rainy days.

If the non-rainy portion is larger, it indicates most days are dry.

2. Pie Chart — Sunshine Category Share

```python
# Create sunshine categories
df['sunshine_category'] = pd.cut(
df['BASEL_sunshine'],
bins=[0,3,7,12],
labels=['Low','Medium','High']
)

sunshine_counts = df['sunshine_category'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(sunshine_counts,
labels=sunshine_counts.index,
autopct='%1.1f%%',
startangle=90)

plt.title("Sunshine Category Distribution")

plt.show()
```
